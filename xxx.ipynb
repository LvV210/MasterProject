```python
import numpy as np
from matplotlib.colors import LinearSegmentedColormap
import matplotlib.pyplot as plt

# Define the colors
space_blue = (0, 0, 50/255)  # Normalizing the blue value to 0-1 range
black = (0, 0, 0)

# Create the colormap
cmap = LinearSegmentedColormap.from_list("space_blue_to_black", [space_blue, black])

# Create a figure with a space-themed color scheme (darker blue and black)
fig, ax = plt.subplots()

# Create a gradient background starting from dark blue to black
ax.set_facecolor('black')
x = np.linspace(0, 1, 500)
y = np.linspace(0, 1, 500)
X, Y = np.meshgrid(x, y)
Z = np.sqrt(X**2 + Y**2)


# Add some stars with varying brightness
N_stars = 50
stars_x = np.random.rand(N_stars) * 1.75
stars_y = np.random.rand(N_stars)
star_sizes = np.random.randint(1, 2, N_stars)
star_colors = np.random.rand(N_stars)
ax.scatter(stars_x, stars_y, color='grey', s=star_sizes, alpha=star_colors)

# # Adjust the color map to dark blue and black
# cmap = plt.get_cmap('Blues')
# cmap = cmap.reversed()


# Remove axes and save the figure
ax.axis('off')
# plt.savefig('/mnt/data/hypervelocity_stars_background_darker_blue_black.png', bbox_inches='tight', pad_inches=0)
ax.imshow(Z, extent=[0, 1.75, 0, 1], origin='lower', cmap=cmap, alpha=1)
fig.savefig("/mnt/c/Users/luukv/Documenten/NatuurSterrkenkundeMasterProject/CodeMP/MasterProject/background.png", dpi=1000)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

# Define the colors
space_blue = (0, 0, 139/255)  # Normalizing the blue value to 0-1 range
black = (0, 0, 0)

# Create the colormap
cmap = LinearSegmentedColormap.from_list("space_blue_to_black", [space_blue, black])

# Display the colormap
gradient = np.linspace(0, 1, 256)
gradient = np.vstack((gradient, gradient))

fig, ax = plt.subplots(figsize=(6, 1))
ax.imshow(gradient, aspect='auto', cmap=cmap)
ax.set_axis_off()

plt.show()
```
