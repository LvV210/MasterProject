```python
import sys
sys.path.append('/mnt/c/Users/luukv/Documenten/NatuurSterrkenkundeMasterProject/CodeMP/MasterProject')
from model_functions import import_SED_models_quickload, select_spectrum
from functions import import_spectra, extract_spectrum_within_range

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
object_list = ['4U1538_52', '4U1700_37', 'LMCX_4', 'SMCX_1', 'VelaX_1', 'CenX_3']
doppler_shift = [2.8, 0.6, -5, -3.5, 0.3, 0]
spectral_lines = [
            [4713.17, r"He I: 4713.17", 4708.5, 4709, 4717.35, 4717.85],
            [4009.26, r"He I: 4009.26", 4006, 4006.5, 4012, 4012.5],
            [5875.66, r"He I: 5875.66", 5867, 5867.5, 5882.5, 5883],
            [4026.21, r"He I: 4026.21", 4018.5, 4019, 4031, 4031.5]
    ]
```

```python
%matplotlib qt
```

```python
fig, axes = plt.subplots(len(object_list), len(spectral_lines), figsize=(20, 25))
for j, system in enumerate(object_list):
    spectra = import_spectra(system)

    for i, line in enumerate(spectral_lines):
        spectrum = select_spectrum(spectra, line[0])
        if spectrum == None:
            axes[j, i].set_axis_off()
            continue
        wav, flux = extract_spectrum_within_range(spectrum[0], spectrum[1], line[0] - 20, line[0] + 20)
        axes[j, i].plot(spectrum[0], spectrum[1])
        axes[j, i].set_title(line[1])
        axes[j, i].set_ylim(min(flux)*0.9, max(flux))
        axes[j, i].set_ylabel(system, fontsize=15)
plt.tight_layout()
plt.show()
```

```python

for j, system in enumerate(object_list):
    fig, axes = plt.subplots(len(spectral_lines), 1, figsize=(12, 7))
    spectra = import_spectra(system)

    for i, line in enumerate(spectral_lines):
        spectrum = select_spectrum(spectra, line[0])
        if spectrum == None:
            axes[i].set_axis_off()
            continue
        wav, flux = extract_spectrum_within_range(spectrum[0], spectrum[1], line[0] - 20, line[0] + 20)
        axes[i].plot(spectrum[0], spectrum[1], color='blue')
        axes[i].set_title(line[1])
        axes[i].set_ylim(min(flux)*0.9, max(flux)*1.1)
        axes[i].set_xlim(min(wav), max(wav))
        axes[i].vlines(line[0] - doppler_shift[j], ymin=min(flux)*0.9, ymax=max(flux)*1.1, color='orange')
        axes[i].vlines(np.array([line[2], line[3], line[4], line[5]]) - doppler_shift[j], ymin=min(flux)*0.9, ymax=max(flux)*1.1, color='green')
        axes[i].set_ylabel(system, fontsize=15)
    plt.suptitle(system)
    plt.tight_layout()
    plt.show()
```
