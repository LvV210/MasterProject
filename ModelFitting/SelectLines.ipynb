```python
def lines_in_range(lines, min_wav, max_wav):
    return [tup for tup in lines if tup[0] > min_wav and tup[0] < max_wav]

def BalmerLines():
    return [[6562.79, "Ha"],
            [4861.35, "Hb"],
            [4340.47, "Hg"],
            [4101.73, "Hd"],
            [3970.08, "He"],
            [3889.06, "Hf"],
            [3835.40, "Hn"]]

def HeliumILines():
    return [(5875.66, r"He I: 5875.66"),
            (4471.50, r"He I: 4471.50"),
            (4026.21, r"He I: 4026.21"),
            (3819.62, r"He I: 3819.62"),
            (3705.02, r"He I: 3705.02"),
            (3634.25, r"He I: 3634.25"),
            (3587.27, r"He I: 3587.27"),
            (4921.93, r"He I: 4921.93"),
            (4387.93, r"He I: 4387.93"),
            (4143.76, r"He I: 4143.76"),
            (4009.26, r"He I: 4009.26"),
            (4713.17, r"He I: 4713.17"),
            (5015.68, r"He I: 5015.68"),
            (5047.74, r"He I: 5047.74")]

def HeliumIILines():
    return [[4686, r"He II: 4686"],
            [5411.53, r"He II: 5411.53"],
            [4199.83, r"He II: 4199.83"]]
```

### Vela X-1

```python
for line in spectral_lines_model_fit['lines']:
    delta = 20
    central_wav = line[0]
    vsini = 250
    min_wav, max_wav = 10000, 0

    # Get spectrum around line
    wav, flux = select_spectrum(spectra, central_wav)
    plt.plot(wav, flux, color='black', linewidth=0.1)
    if min(wav) < min_wav:
        min_wav =  min(wav)
    if max(wav) > max_wav:
        max_wav = max(wav)
    wav, flux = extract_spectrum_within_range(wav, flux, central_wav - delta, central_wav + delta)

    # # Use a smoothing spline to fit the continuum
    # spline = UnivariateSpline(wav, flux, s=0.01 * len(wav))  # Adjust 's' for smoothness
    # continuum = spline(wav)
    # Normalize the spectrum
    # flux = flux / continuum
    plt.plot(wav, flux, color='black')

    # Get models around line
    # for key, model in models.items():
    #     wav, flux = extract_spectrum_within_range(np.array(model['WAVELENGTH']), np.array(model['FLUX']), central_wav - delta, central_wav + delta)
    #     # Apply doppler broadening
    #     # wav, flux = pyasl.equidistantInterpolation(wav, flux, "2x")
    #     # flux = pyasl.rotBroad(wav, flux, 0.0, vsini)
    #     plt.plot(wav, flux)

    plt.vlines(x=[line[2], line[5]], ymin=0, ymax=2, color='midnightblue')
    plt.vlines(x=[line[3], line[4]], ymin=0, ymax=2, color='darkred')

    plt.title(f"{line[1]}")
    plt.ylim(0,1.25)

balmer = lines_in_range(BalmerLines(), min_wav, max_wav)
heliumI = lines_in_range(HeliumILines(), min_wav, max_wav)
heliumII = lines_in_range(HeliumIILines(), min_wav, max_wav)

for line in balmer:
    plt.text(line[0], 1.2, line[1], ha='center', va='bottom', rotation=90, size=7)

    delta = 20
    central_wav = line[0]

    # Get spectrum around line
    wav, flux = select_spectrum(spectra, central_wav)
    plt.plot(wav, flux, color='black', linewidth=0.1)
    wav, flux = extract_spectrum_within_range(wav, flux, central_wav - delta, central_wav + delta)

    for key, model in models.items():
        wav, flux = extract_spectrum_within_range(np.array(model['WAVELENGTH']), np.array(model['FLUX']), central_wav - delta, central_wav + delta)
        plt.plot(wav, flux)

for line in heliumI:
    plt.text(line[0], 1.2, line[1], ha='center', va='bottom', rotation=90, size=7)

    delta = 20
    central_wav = line[0]

    # Get spectrum around line
    wav, flux = select_spectrum(spectra, central_wav)
    plt.plot(wav, flux, color='black', linewidth=0.1)
    wav, flux = extract_spectrum_within_range(wav, flux, central_wav - delta, central_wav + delta)

    for key, model in models.items():
        wav, flux = extract_spectrum_within_range(np.array(model['WAVELENGTH']), np.array(model['FLUX']), central_wav - delta, central_wav + delta)
        plt.plot(wav, flux)

for line in heliumII:
    plt.text(line[0], 1.2, line[1], ha='center', va='bottom', rotation=90, size=7)

    delta = 20
    central_wav = line[0]

    # Get spectrum around line
    wav, flux = select_spectrum(spectra, central_wav)
    plt.plot(wav, flux, color='black', linewidth=0.1)
    wav, flux = extract_spectrum_within_range(wav, flux, central_wav - delta, central_wav + delta)

    for key, model in models.items():
        wav, flux = extract_spectrum_within_range(np.array(model['WAVELENGTH']), np.array(model['FLUX']), central_wav - delta, central_wav + delta)
        plt.plot(wav, flux)

plt.show()
```
