#### Function to make a markdown table of a pandas dataframe

```python
def display_df(df2: pd.DataFrame, scientific_notation: bool = True, power: int = False, round_: bool = False, round_digits:int=1, return_md_text:bool=False, latex:bool=False, error_columns:list=False):
    """
    Displays a given dataframe in a nice way with better collumn names and units.
    """
    df = df2.copy()
    df_save = df2.copy()

    columns = {"R_true": (r"$R_{true}[R_{\odot}]$"),
               "R_true_err": r"$\delta R_{true} [R_{\odot}]$",
                "R_expected": r"$R_{expected}[R_{\odot}]$",
                "R_expected_err_high": r"$\delta_{high} R_{expected} [R_{\odot}]$",
                "R_expected_err_low": r"$\delta_{low} R_{expected} [R_{\odot}]$",
                "R_true/R_expected": r"$\frac{R_{true}}{R_{expected}}$",
                "R_ST_err": r"$\delta R_{ST} [R_{\odot}]$",
                "R_ST": r"$R_{ST} [R_{\odot}]$",
                "L_true": r"$L_{true}[L_{\odot}]$",
                "L_true_err_low": r"$\delta_{low} L_{true}[L_{\odot}]$",
                "L_true_err_high": r"$\delta_{high} L_{true}[L_{\odot}]$",
                "L_expected": r"$L_{expected}[L_{\odot}]$",
                "L_true/L_expected": r"$\frac{L_{true}}{L_{expected}}$",
                "Mopt": r"$M_{obs} [M_{\odot}]$",
                "Mopt_err": r"$\delta M_{obs} [M_{\odot}]$",
                "M_exp": r"$M_{BONN} [M_{\odot}]$",
                "M_exp_err": r"$\delta M_{BONN} [M_{\odot}]$",
                "M_ST": r"$M_{ST} [M_{\odot}]$",
                "M_ST_err": r"$\delta M_{ST} [M_{\odot}]$",
                "M_HRD": r"$M_{HRD} [M_{\odot}]$",
                "M_HRD_err": r"$\delta M_{HRD} [M_{\odot}]$",
                "ST": "Spec. Type",
                "ST_short": "Spec. Type (short)",
                "period": r"$P[days]$",
                "spinperiod": r"$P_{spin}[s]$",
                "eclipseduration": r"eclipseduration[units]",
                "RV": r"$v_{rad}[km s^{-1}]$",
                "Mob": r"$M_{ob}[M_{\odot}]$",
                "Mx": r"$M_{x} [M_{\odot}]$",
                "Rob": r"$R_{ob}[R_{\odot}]$",
                "Rx": r"$R_{x}[R_{\odot}]$",
                "parallax": r"$p'' ['']$",
                "errparallax": r"$\delta p''['']$",
                "distance": r"$d[pc]$",
                "distanceBJ": r"$d_{BJ}[pc]$",
                "luminosity": r"$L_{ob}[L_{\odot}]$",
                "(B-V)obs": r"$(B-V)_{obs}$",
                "(B-V)0": r"$(B-V)_{0}$",
                "BC": r"$BC_{v}$",
                "mv": r"$m_{v}$",
                "Teff": r"$T_{eff}[K]$",
                "Mv": r"$M_{v}$",
                "log(g_spec)": r"$log(g_{spec})$",
                "log(L/Lsun)": r"$log(\frac{L}{L_{\odot}}$",
                "M_spec": r"$M_{spec}$"}
    
    columns = { "R_true": r"$R_{true}[R_{\odot}]$",
                "Ropt": r"$R_{opt}[R_{\odot}]$",
                "R_expected": r"$R_{expected}[R_{\odot}]$",
                "R_ST": r"$R_{ST} [R_{\odot}]$",
                "R": r"$R_{spec} [R_{\odot}]$",

                "L_true": r"$L_{JH}[L_{\odot}]$",
                "LBV_true": r"$L_{BV} [L_{\odot}]$",
                "LBV_final": r"$L_{BV} [L_{\odot}]$",
                "L_ST": r"$L_{expected}[L_{\odot}]$",
                "log(L/Lsun)": r"$^{10}$ log $(\frac{L}{L_{\odot}})$",

                "Mopt": r"$M_{obs} [M_{\odot}]$",
                "M_exp": r"$M_{BONN} [M_{\odot}]$",
                "M_ST": r"$M_{ST} [M_{\odot}]$",
                "M_HRD": r"$M_{HRD} [M_{\odot}]$",
                "M_spec": r"$M_{spec} [M_{\odot}]$",

                "Teff": r"$T_{eff}$ [K]",
                "Tmodel": r"$T_{model}$ [K]",
                "RL_fal": r"$R_{L,fal} [R_{\odot}]$",

                "RL_q": r"$R_{L,q} [R_{\odot}]$",
                "RL_g": r"$R_{L,g} [R_{\odot}]$",

                "Mv": r"$M_{v}$",
                "(J-H)0": r"$(J-H)_{0}$",
                "(B-V)0": r"$(B-V)_{0}$",
                "Mh": r"$M_{h}$",
                "BCh": r"$BC_{h}$",
                "BCv": r"$BC_{v}$",

                "r_med_photogeo": r"$d [pc]$",

                "X_rayFlux_SWIFTXRT": r"$F_{swift, X-ray}$",

                # Falenga table
                "Orbital period": r"$P_{orb}$",
                "Orbital period error": r"$\delta P_{orb}$",
                "ax sin i": r"$a_{x} \cdot sin(i)$",
                "ST": "Spec. Type"
                }

    if power:
        # Identify columns with numerical values
        numeric_cols = df.select_dtypes(include=[np.number]).columns
        # Scale down
        for col in numeric_cols:
            df[col] /= 10**power
            df_save[col] /= 10**power

    # Scientific notation
    if scientific_notation:
        df = df.map(lambda x: "*" + "%.2E" % x if isinstance(x, (int, float, np.floating)) else x)
        df_save = df_save.map(lambda x: "*" + "%.2E" % x if isinstance(x, (int, float, np.floating)) else x)


    # Round
    if round_:
        df = df.map(lambda x: round(x, round_digits) if isinstance(x, (int, float, np.floating)) else x)
        df_save = df_save.map(lambda x: round(x, round_digits) if isinstance(x, (int, float, np.floating)) else x)

    if error_columns:
        # Write error columns like 1
        for error_column in error_columns:
            # For asymmetric errors
            if len(error_column) == 3:
                # Calculate the formatted error string for each row
                error_strings = df_save.apply(lambda row: f"${row[error_column[0]]}^{{+{row[error_column[1]]}}}_{{-{row[error_column[2]]}}}$", axis=1)

                # Create a new column in the DataFrame with the formatted error strings
                df[error_column[0]] = error_strings

            # For symmetric errors
            if len(error_column) == 2:
                # Calculate the formatted error string for each row
                error_strings = df_save.apply(lambda row: f"${row[error_column[0]]} \pm {{{row[error_column[1]]}}}$", axis=1)
                
                # Create a new column in the DataFrame with the formatted error strings
                df[error_column[0]] = error_strings

            # Filter the list of columns to remove to only include those that exist in the DataFrame
            existing_columns_to_remove = [col for col in error_column[1:] if col in df.columns]
            # Remove the error columns
            df.drop(columns=existing_columns_to_remove, inplace=True)

    # Rename columns
    df = df.rename(columns=columns)

    # Also add the power divided out to the column name
    if power:
        # Update column names to reflect the scaling
        df.columns = [(r'$' + col.replace('$', '') + rf"(10^{{{power}}})$") if col != 'id' else col for col in df.columns]

    # To markdown
    markdown_text = df.to_markdown(index=False, tablefmt="pipe")

    # Remove '*' from text
    markdown_text = markdown_text.replace('*', '')

    if latex:
        format_string = "{:." + str(round_digits) + "f}"
        return print(df.to_latex(index=False, float_format=format_string.format, na_rep='--'))
    if return_md_text:
        return markdown_text
    else:
        return md(markdown_text)
```

Column names:
- id:           Name of the object
- ST:           Spectral type <br><br>
- Mopt:         Mass based on orbital parameters
- M_exp:        Mass determined with the BONNSAI models
- M_HRD:        Mass determined with the Chen 2015 evolutionary tracks
- M_ST:         Mass based on martins models and ST <br><br>
- Ropt:         Radius based on $\theta_{e}$ **=R_true**
- R_expected:   From L=4piR2sigmaT4
- R_ST:         Radius based on Martins models and spectral type <br><br>
- L_true:       Photometric luminosity
- L_ST:         Luminosity based on martins models <br><br>
- Teff:         Effective temperature based on martins models and ST
- Teff_mod:     Model Teff <br><br>
- P_orbit:      Orbital period
- P_spin:       Spin period of compact object
- Mx:           Mass of compact object
